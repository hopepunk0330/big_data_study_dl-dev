```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# 회귀 모델 실습
from sklearn.model_selection import train_test_split
# from sklearn.metrics import *
# 회귀 모델의 평가 지표를 import 한다
from sklearn.metrics import root_mean_squared_error
from sklearn.metrics import r2_score # 회귀 모델의 대표저인 성능 평가 지표 사용된다.
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_absolute_percentage_error 
# 정규화, 표준화 
from sklearn.preprocessing import StandardScaler, MinMaxScaler
```

```python
# 1. 비지니스 이해 2. 데이터 이해, 3. 데이터 전처리 4. 모델링, 5. 평가, 6. 배포
# 3. 데이터 전처리
# 3.1 데이터 준비 : https://bit.ly/advertising_csv
path = ''
data1 = pd.read_csv('../data/advertising.csv') # csv 파일 읽어서 데이터프레임 생성
data1.head()
```

```python
# 3.2 데이터 분리 : 데이터와 레이블 분리
target = 'Sales'
# 데이터 추출
# X = data1.drop(target, axis=1) # Sales columns 삭제하고 데이터 대입
X = data1.iloc[:,1:-1]
# 레이블 추출
y = data1.loc[:,'sales']
```

```python
# 3.3 데이터 분할 : 학습, 테스트 데이터 분할
X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=.3, random_state=20)
```

```python
# 3.4 표준화 : 0~1 사이 값으로 변환, MinMaxScaler
scaler = MinMaxScaler()
# fit() :  (x - min) / (max-min)식에 필요한 학습데이터 컬럼별 min, max 값을 구한다.
# scaler.fit(X_train) # (x - min) / (max-min) : x 데이터 1개
# 변환 : (x - min) / (max-min)
# X_train_scaled = scaler.transform(X_train) 
# 학습데이터에 대한 변환
X_train_scaled = scaler.fit_transform(X_train)
# 테스트 데이터에 대한 변환 
X_test_scaled = scaler.transform(X_test)
```

```python
# 4. 모델링 : 학습(fit()) -> 예측값(predict()) -> 평가(eval)
# 4.1 모델 선택 : LinearRegression (단순선형회귀 모델)
from sklearn.linear_model import LinearRegression
model = LinearRegression() # 아무것도 없다. 기본값으로 처리하겠다란 의미
# 4.2 학습
model.fit(X_train_scaled, y_train)
# 4.3 예측값 추출
preds = model.predict(X_test_scaled) # 테스트데이터(새로운 데이터)에 대한 예측값
```

```python
# 평가
print(f'RMSE : {root_mean_squared_error(y_test, preds)}')
print(f'R2 :  {r2_score(y_test, preds)}')
print(f'MAE : {mean_absolute_error(y_test, preds)}')
```

```python
# R2(결정계수, Coefficent of Determination)
# 회귀 모델이 실제 데이터를 얼마나 잘 설명하는지 나타내는 통계적 지표
# 0 ~ 1
# 1 : 모델이 데이터를 완벽하게 설명한다.
# 0 : 단순히 데이터의 평균값으로 예측하는 것과 차이가 없다.
# 음수 : 설명 못한다.
```
