```python
### 광고비에 따른 매출액 예측 ###


# DataLoader
path = 'https://bit.ly/advertising_csv'
adv = pd.read_csv(path)
```

```python
target = 'Sales'                # 원래답
x = adv.drop(target, axis=1)    # 원래답 제거 => 데이터만 남겠다
y = adv.loc[:, target]          # 원래답 추출
x_np = x.values                 # 2차원 ndarray
y_np = y.values                 # 1차원 ndarray
print(type(x_np))
print(x_np.shape)
print(type(y_np))
print(y_np.shape)
```

```python
# ⭐⭐⭐ 넘파이 배열 x, y를 텐서로 변환


x_tensor = torch.tensor(x_np, dtype=torch.float32)
y_tensor = torch.tensor(y_np, dtype=torch.float32)
print(x_tensor.shape, y_tensor.shape)
```

```python
# 데이터와 답을 묶는 작업

from torch.utils.data import DataLoader, TensorDataset
dataset = TensorDataset(x_tensor, y_tensor)
print(dataset[:3])
print(len(dataset))
```

```python
# 데이터 로더 생성

data_loader = DataLoader(dataset, batch_size = 32, shuffle = True)
```

```python
# 데이터 셋 실행 : 내적함 -> ⭐⭐결과의 한 줄씩 신경망에 들어감

for n, batch in enumerate(data_loader):
    x_batch, y_batch = batch
    print(f'배치 {n}, x 배치 크기:{x_batch.shape}')
```

```python
# 디바이스 할당

# 1. 데이터 준비


# 라이브러리 로딩
import torch
from torch import nn
import numpy as np
import pandas as pd


# 데이터 로딩 및 x, y 분할
path = 'https://bit.ly/advertising_csv'
adv = pd.read_csv(path)


target = 'Sales'
x = adv.drop(target, axis=1)
y = adv.loc[:, target]


# numpy로 변환한 다음 텐서로 변환
x_np = x.values
y_np = y.values
x_tensor = torch.tensor(x_np, dtype=torch.float32)
y_tensor = torch.tensor(y_np, dtype=torch.float32)
```

```python
# 디바이스 준비 (GPU로 할당)
device = 'cuda' if torch.cuda.is_available() else 'cpu'
print(f'Using {device} device.')
```

```python
# 3. 모델과 데이터셋을 디바이스 할당


# 📌모델을 디바이스(GPU)에 할당하기

model = nn.Sequential( nn.Linear(3, 1) )
model = model.to(device)    # 모델안에 가중치(값 저장소)가 있다
```

```python
# 📌 데이터 세트를 디바이스에 할당.
x, y = x_tensor.to(device), y_tensor.to(device)
```

```python
# 디바이스 확인

print(next(model.parameters()).device)
print(x.device, y.device)   # 데이터의 목록이 어디에 들어갔는지 확인 : GPU 갯수별로 번호를 쓸수 있고 원하는 장치로 보낼수 있음
```

```python
# GPU -> CPU 로 이동

x_cpu = x.cpu()
y_cpu = y.to('cpu')
print(x_cpu.device, y_cpu.device)
```
